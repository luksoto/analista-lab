## Demostración · Limpieza reproducible

La venta repetida es una copia exacta. No habría que eliminar una clave repetida con valores contradictorios sin investigar. El coste ausente **sigue siendo desconocido**: no se rellena con cero ni con el original limpio.


```python
df = raw.copy()
repetidas = df[df["venta_id"].duplicated(keep=False)]
conflictos = repetidas.groupby("venta_id").nunique(dropna=False).gt(1).any(axis=1)
assert not conflictos.any(), "Investigar ventas repetidas con valores contradictorios"
df["ciudad"] = df["ciudad"].str.strip().str.title()
df["coste"] = pd.to_numeric(df["coste"], errors="coerce")
df = df.drop_duplicates(subset=["venta_id"]).copy()
df["fecha"] = pd.to_datetime(df["fecha"], format="%Y-%m-%d")
df["mes"] = df["fecha"].dt.to_period("M").astype(str)
df["margen"] = df["importe"] - df["coste"]
print("Filas limpias:", len(df))
print("Costes conocidos:", df["coste"].notna().sum(), "/", len(df))
print("Ingresos totales:", df["importe"].sum())
print("Margen de operaciones con coste conocido:", df.loc[df["coste"].notna(), "margen"].sum())
print("No es el margen total completo: falta recuperar un coste.")
```
