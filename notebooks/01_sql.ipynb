Objetivos: filtrar, agrupar, unir y calcular acumulados. Dialecto: SQLite. Las funciones de fecha y LIMIT pueden diferir en otros motores.


```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
setup_sql = "DROP TABLE IF EXISTS ventas;\nCREATE TABLE ventas (venta_id INTEGER PRIMARY KEY, fecha TEXT, cliente_id INTEGER, ciudad TEXT, categoria TEXT, canal TEXT, importe REAL, coste REAL);\nINSERT INTO ventas VALUES\n(1,'2026-01-05',1,'Neuquén','Hogar','Web',120,75),\n(2,'2026-01-06',2,'Córdoba','Tecnología','Tienda',300,210),\n(3,'2026-01-12',3,'Mendoza','Hogar','Web',80,45),\n(4,'2026-01-16',1,'Neuquén','Tecnología','Tienda',250,180),\n(5,'2026-02-02',4,'Córdoba','Hogar','Web',140,85),\n(6,'2026-02-07',3,'Mendoza','Tecnología','Tienda',320,230),\n(7,'2026-02-15',5,'Neuquén','Hogar','Web',100,60),\n(8,'2026-02-22',2,'Córdoba','Tecnología','Web',200,145),\n(9,'2026-03-04',6,'Mendoza','Hogar','Tienda',160,100),\n(10,'2026-03-08',5,'Neuquén','Tecnología','Tienda',280,195),\n(11,'2026-03-18',4,'Córdoba','Hogar','Tienda',90,55),\n(12,'2026-03-26',6,'Mendoza','Tecnología','Web',400,280);\nCREATE TABLE clientes (cliente_id INTEGER PRIMARY KEY, segmento TEXT);\nINSERT INTO clientes VALUES (1,'Frecuente'),(2,'Frecuente'),(3,'Frecuente'),(4,'Ocasional'),(5,'Ocasional'),(6,'Ocasional');\n"
conn.executescript(setup_sql)
print("SQLite", sqlite3.sqlite_version)
print(pd.read_sql_query("SELECT * FROM ventas LIMIT 5", conn))
```

## Comprueba la base

Hay 12 ventas y 6 clientes distintos. Ingresos: 2440 UM; costes: 1660 UM; margen: 780 UM. Estos totales pertenecen al archivo **limpio**, con todos los costes conocidos.


```python
assert conn.execute("SELECT COUNT(*) FROM ventas").fetchone()[0] == 12
assert conn.execute("SELECT SUM(importe) FROM ventas").fetchone()[0] == 2440
assert conn.execute("SELECT SUM(importe-coste) FROM ventas").fetchone()[0] == 780
print("Base preparada y totales comprobados.")
```
